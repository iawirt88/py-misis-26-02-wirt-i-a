```python
stat_age = df["Age"].agg(["mean", "median", "min", "max", "std", "count"])
stat_age
```

```python
age_median = stat_age["median"]
df["Age"] = df["Age"].fillna(age_median)
```

```python
print(df["Age"].tail())
stat_age_filled = df["Age"].agg(["mean", "median", "min", "max", "std", "count"])
stat_age_filled
```

```python
comprasion_age = pd.DataFrame({
    "До обработки" : stat_age,
    "После обработки" : stat_age_filled
})
comprasion_age
```

```python
df = df.dropna(subset=["Embarked"])
df["Embarked"].isna().sum()
```

```python
if df["Cabin"].isna().mean() * 100 > 50: df = df.drop(columns="Cabin")
df.columns
```

```python
df.isna().sum()
```

```python
print({"До обработки" : (col, rows), "После обработки" : df.shape})
```

```python
np.mean(df["Survived"])
```

```python
man_surv = np.mean(df["Survived"].where(df["Sex"] == "male"))
woman_surv = np.mean(df["Survived"].where(df["Sex"] == "female"))
print("Выживаемость мужчин: ", man_surv, '\n', "Выживаемость женщин: ", woman_surv)
```

```python
data = [
    [np.mean(df["Age"].where((df["Sex"] == "male") & (df["Survived"] == 1))),
    np.mean(df["Age"].where((df["Sex"] == "male") & (df["Survived"] == 0)))],
    [np.mean(df["Age"].where((df["Sex"] == "female") & (df["Survived"] == 1))),
    np.mean(df["Age"].where((df["Sex"] == "female") & (df["Survived"] == 0)))]
]
print("Срдений возраст выживших/погибших")
pd.DataFrame(
    data, 
    columns=[ "Выжили", "Погибли" ],
    index=[ "Мужчины", "Женщины" ]
)
```

```python
most_payed_male = df.loc[(df["Age"] > 30) & (df["Pclass"] == 1)]
print(most_payed_male.shape[0])
most_payed_male.head(10)
```

```python
df.loc[(df["Age"] <= 40) & (df["Age"] >= 20) & (df["Embarked"] == "S")]
```

```python
group_by_sex_cls = df.groupby(["Pclass", "Sex"]).agg({
    "PassengerId" : "count",
    "Age" : "mean",
    "Survived" : "mean",
    "Fare" : "mean"
}).reset_index()
group_by_sex_cls
```

```python
ports_surv_stat = df.groupby(["Embarked"]).agg({
    "PassengerId" : "count",
    "Survived" : "mean"
})
ports_surv_stat
```

```python
df.groupby(["Pclass"]).agg({"Survived" : "mean"}).idxmax()["Survived"]
```

```python
df.to_csv("titanic_processed.csv", index=False, sep=",")
group_by_sex_cls.to_csv("titanic_statistics.csv", index=False, sep=",")
```
